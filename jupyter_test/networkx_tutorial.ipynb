```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
G.add_edge(1,2)
G.add_edge(2,3, weight=0.9)

nx.draw_spring(G, with_labels=True)
plt.show()
```

```python
G = nx.complete_graph(5)
nx.draw_spring(G, with_labels=True)
plt.show()
```

```python
edge_list = [(1, 2), (1, 3), (1, 4), (2, 3), (2, 4), (3, 4)]
G = nx.Graph()
G = nx.from_edgelist(edge_list)
nx.draw_planar (G, with_labels=True)
plt.show()
```

```python
edge_list = [(1, 2), (2, 3), (3, 4), (3, 5), (4, 6), (6, 7)]
G = nx.from_edgelist(edge_list)

nx.draw_spring(G, with_labels=True)
plt.show()

nx.draw_circular(G, with_labels=True)
plt.show()

nx.draw_shell(G, with_labels=True)
plt.show()

nx.draw_spectral(G, with_labels=True)
plt.show()

nx.draw_random(G, with_labels=True)
plt.show()

nx.draw_planar (G, with_labels=True)
plt.show()
```

```python
print(nx.adjacency_matrix(
    G
))
```

```python
import networkx as nx

# Crear un MultiDiGraph
G = nx.MultiDiGraph()

# Añadir nodos
G.add_node("A")
G.add_node("B")

# Añadir múltiples aristas entre A y B
G.add_edge("A", "B", key=0, weight=3)
G.add_edge("A", "B", key=1, weight=5)
G.add_edge("B", "A", weight=2)  # Dirección opuesta
nx.draw_spring(G, with_labels=True)
# Mostrar aristas
print(list(G.edges(data=True, keys=True)))
```

```python
# Jerarquía básica
G.add_edge("Fruta", "Ingrediente", key="subclass_of", relation="subclass_of")
G.add_edge("Lácteo", "Ingrediente", key="subclass_of", relation="subclass_of")
G.add_edge("Endulzante", "Ingrediente", key="subclass_of", relation="subclass_of")
G.add_edge("Hielo", "Ingrediente", key="subclass_of", relation="subclass_of")
nx.draw_circular(G, with_labels=True)
```
